```python
#accept a sequence of comma separated numbers from user and genarate a tuple with those numbers and find avarage of all items in that tuplw
number=input("Enter the numbers separated with commas:")
num_t=tuple(int(x) for x in number.split(","))
average=sum(num_t)/len(num_t)
print("The average is:",average)
```
