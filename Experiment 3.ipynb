```python
#Reverse a given string
s=input("Enter a string:")
new_string=s[-1]+s[1:-1]+s[0]
print("New string is:",new_string)
```

```python
#string made of first 2 and the last 2 chars from given a string
s=input("Enter a string:")
new_string=s[0:3]+s[-2::1]
print("New string is:",new_string)
```

```python
#create a single string separated by swapping the character at position 1.
s1=input("Enter the first string:")
s2=input("Enter the second string:")
new_s1=s2[0]+s1[1:]
new_s2=s1[0]+s2[1:]
print("New string is:",new_s1,new_s2)
```

```python
#get a string from a given string where all occurrences of its first char have been changed to$, except the first char itself

s=input("Enter the string:")
new_string=s[0]+s[1::1].replace(s[0],"s")
print("New string is :",new_string)
```
