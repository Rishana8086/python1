```python
#include list of colors and separate them with comma
colors=input("Enter the colors")
color_list=(colors.split(","))
print("The first color is:",color_list[0])
print("The last color is:",color_list[-1])

```

```python
#input extension from file name
filename=input("Enter the filename:")
extension=filename.split(",")[-1]
print("the extension is:",extension)
```

```python
#Accept sequence of comma separated numbers feom user and genarate a list and find sum
number=input("Enter the numbers:")
num2=[int(x) for x in number.split(",")]
print("the list is:",num2)
print("sum of num",sum(num2))
```
