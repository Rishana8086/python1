```python
#accept 5 students roll no and name and sort them in ascending and descending order 
print("enter 5 roll number of students:")
roll=input().split(",")
print("Enter 5 names of students:")
name=input().split(",")
student=(dict(zip(map(int,roll),name))
ascending=dict(sorted(student.items()))
descending=dict(sorted(student.items(),reverse=True))
print("The dictionary is:",student)
print("Ascending order :",acsending)
print("Descending order:",descending)
```
